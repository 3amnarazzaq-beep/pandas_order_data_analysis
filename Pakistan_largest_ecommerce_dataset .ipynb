```python
# combine categories
orders_part1 = orders[orders["category_name_1"] == "Electronics"]
orders_part2 = orders[orders["category_name_1"] == "Fashion"]
combined_products = pd.concat([orders_part1, orders_part2])

print(combined_products)
```

```python
# combine horizontally
order1 = orders[["item_id", "price"]]
order2 = orders[["qty_ordered", "payment_method"]]

combined_orders = pd.concat([order1, order2], axis=1)

print(combined_orders.head())
```

```python
customers = orders[['Customer ID']].drop_duplicates()
orders1 = orders.copy()
print(customers.head())
print(orders1.head())
```

```python
# inner merge
inner_merged = pd.merge(customers, orders1, on='Customer ID', how='inner')

print(inner_merged)
print("Number of records:", len(inner_merged))
```

```python
# left merge
left_merged = pd.merge(customers, orders1, on='Customer ID', how='left')

print(left_merged)
print("Number of records:", len(left_merged))
```

```python
# compare merges
print("Inner merge rows:", len(inner_merged))
print("Left merge rows:", len(left_merged))

print("Records not matched:", len(left_merged) - len(inner_merged))
```

```python
# join customer and df1 data
# Join customer and order data
customer_data = customers.set_index('Customer ID')
order_data = orders1.set_index('Customer ID')

joined_data = customer_data.join(order_data, how='inner')

print(joined_data)
```

```python
# Merge different column names
customer_data = orders[['Customer ID']].drop_duplicates()
order_data = orders[['Customer ID', 'item_id']].copy()

customer_data = customer_data.rename(columns={'Customer ID': 'customer_id'})
order_data = order_data.rename(columns={'Customer ID': 'customer_key'})

merged_data = pd.merge(
    customer_data,
    order_data,
    left_on='customer_id',
    right_on='customer_key'
)

print(merged_data)
```

```python
# payment methods
print(orders['payment_method'].unique())

payment_counts = orders['payment_method'].value_counts()

print(payment_counts)
print("Highest:", payment_counts.idxmax())
```

```python
# product categories
print(orders['category_name_1'].unique())

category_analysis = orders.groupby('category_name_1').agg(
    Orders=('category_name_1', 'count'),
    Quantity_Sold=('qty_ordered', 'sum'),
    Average_Price=('price', 'mean')
)

print(category_analysis)
```

```python
# Create business report
business_report = category_analysis.reset_index()

business_report.columns = [
    'Category',
    'Orders',
    'Quantity Sold',
    'Average Price'
]

print(business_report)
```
